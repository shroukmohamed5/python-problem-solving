## Problem 1: Separate Letters and Numbers

Ask the user for a string that contains both letters and numbers (e.g., "Py2th0on3"). Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the sum of the numbers found and join the letters to form a word.

**Example:**
- Input: "Py2th0on3"
- Output: Word: "Python", Sum: 5

````python
text=input("enter a string :")
number=[]
letters=[]
for c in text:
    if c.isalpha():
        letters.append(c)
    elif c.isdigit():
        number.append(int(c))
            
word="".join(  letters)
total=sum(number)
print("letters",letters)
print("sum",total)
print("word", word)

        
````

## Problem 2: Password Validation

Ask a user to enter a password. Check if it is valid based on these rules:
- Length must be between 8 and 16 characters
- It must contain at least one uppercase letter
- It must contain at least one special character (e.g., @, #, $)

**Example:**
- Input: "Guest@123"
- Output: "Password is Secure."

````python
password=input("enter your password :")

has_upper=False
has_special=False

for char in password:
    if char.isupper():
        has_upper=True
    if char in "@#%$":
        has_special=True
if 8<=len(password)<=16 and has_special and has_upper:
        print("Password is Secure.")
else:
    print("Password is Not Secure.")
        

 
````

## Problem 5: Student Test Averages

Create a dictionary where keys are student names and values are lists of 3 test scores. Write a program that calculates the average for each student. If the average is > 90, print the student's name followed by "Honor Roll."

**Example:**
- Input: {'Sam': [95, 90, 92]}
- Output: Sam: Average 92.3 - Honor Roll

````python
students = {
    "Sam": [95, 90, 92],
    "Ali": [80, 85, 88],
    "Sara": [100, 95, 98]
}

for name, scores in students.items():
    average = sum(scores) / len(scores)

    if average > 90:
        print(f"{name}: Average {average:.1f} - Honor Roll")
    else:
        print(f"{name}: Average {average:.1f}")
````

- Input: word1 = "abc", word2 = "xyz"

## Problem 7: Digital Root Calculation

Ask for a multi-digit number. Sum its digits. If the result is more than one digit, keep summing the digits until you get a single-digit result.

**Example:**
- Input: 98
- Output: Digital Root: 8 (9+8=17, 1+7=8)

````python
num = input("Enter a number: ")

while len(num) > 1:
    total = 0
    for digit in num:
        total += int(digit)
    num = str(total)

print("Digital Root:", num)
````

## Problem 8: Element Frequency Counter

Given a list of numbers with many duplicates, return a list of tuples showing each unique element and its frequency (how many times it appeared).

**Example:**
- Input: [1, 1, 2, 3, 3, 3, 4]
- Output: [(1, 2), (2, 1), (3, 3), (4, 1)]

````python
numbers = [1, 1, 2, 3, 3, 3, 4]

result = []

for num in numbers:
    if num not in [x[0] for x in result]:
        result.append((num, numbers.count(num)))

print(result)
````

## Problem 9: Factorial and Sum Divisibility

Ask the user for a number $n$. Calculate the factorial (n!) using a loop, and then check if the resulting factorial is divisible by the sum of numbers from 1 to $n$.

**Example:**
- Input: 4
- Output: Factorial is 24, Sum is 10. 24 is not divisible by 10.

````python
n = int(input("Enter a number: "))

factorial = 1
for i in range(1, n + 1):
    factorial *= i

total = sum(range(1, n + 1))

print("Factorial is", factorial)
print("Sum is", total)

if factorial % total == 0:
    print(factorial, "is divisible by", total)
else:
    print(factorial, "is not divisible by", total)
````

## Problem 10: Shopping Cart System

Start with a dictionary of items and prices: {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}. Allow the user to "Add" or "Remove" items from a cart. If they add an item not in the store, print an error. Finally, print the total bill.

**Example:**
- Input: Add 'apple', Add 'milk'
- Output: Total Bill: $2.0

````python
store = {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}

cart = []

item1 = input("Add item: ")

if item1 in store:
    cart.append(item1)
else:
    print("Item not found")

item2 = input("Add item: ")

if item2 in store:
    cart.append(item2)
else:
    print("Item not found")

total = 0

for item in cart:
    total += store[item]

print("Total Bill: $", total)
````

## Problem 11: Sentence Palindrome Check

Check if a full sentence is a palindrome, ignoring spaces and capitalization.

**Example:**
- Input: "Race car"
- Output: True

````python
sentence = input("Enter a sentence: ")

clean = sentence.replace(" ", "").lower()

if clean == clean[::-1]:
    print(True)
else:
    print(False)
````

## Problem 13: Find Peaks in List

In a list of numbers, find all "peaks." A peak is a number that is greater than the number to its left and the number to its right.

**Example:**
- Input: [1, 5, 2, 6, 3]
- Output: Peaks: [5, 6]

````python
numbers = [1, 5, 2, 6, 3]

peaks = []

for i in range(1, len(numbers) - 1):
    if numbers[i] > numbers[i-1] and numbers[i] > numbers[i+1]:
        peaks.append(numbers[i])

print("Peaks:", peaks)
````

## Problem 14: Flip Dictionary with Duplicate Values

Take a dictionary like {'A': 1, 'B': 2, 'C': 1} and flip it. If two keys have the same value, the new value should be a list of those keys.

**Example:**
- Input: {'A': 1, 'B': 2, 'C': 1}
- Output: {1: ['A', 'C'], 2: ['B']}

````python
data = {'A': 1, 'B': 2, 'C': 1}

result = {}

for key, value in data.items():
    if value not in result:
        result[value] = []
    result[value].append(key)

print(result)
````

## Problem 15: Vowel Index Mapper

Create a dictionary where the keys are the vowels found in a user's string and the values are lists of the indices where those vowels appear.

**Example:**
- Input: "banana"
- Output: {'a': [1, 3, 5]}

````python
text = input("Enter a string: ")

result = {}

for i in range(len(text)):
    if text[i] in "aeiou":
        if text[i] not in result:
            result[text[i]] = []
        result[text[i]].append(i)

print(result)
````
